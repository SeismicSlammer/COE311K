```python
import random
import time
import numpy as np
import matplotlib.pyplot as plt

#Naive matrix multiplication
def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

#Generate random matrix
def generate_matrix(n):
    return [[random.random() for _ in range(n)] for _ in range(n)]

#Matrix sizes (3, 6, 12, ..., 384)
sizes = [3 * (2**i) for i in range(8)]

naive_times = []
numpy_times = []

for n in sizes:

    # Generate matrices
    A = generate_matrix(n)
    B = generate_matrix(n)


    start = time.time()
    naive_matrix_multiplication(A, B)
    end = time.time()
    naive_times.append(end - start)

    #Convert to numpy
    A_np = np.array(A)
    B_np = np.array(B)

    #Time numpy implementation
    start = time.time()
    np.dot(A_np, B_np)
    end = time.time()
    numpy_times.append(end - start)

#Plot results
plt.figure()
plt.plot(range(1, 9), naive_times, marker='o', label='Naive')
plt.plot(range(1, 9), numpy_times, marker='o', label='NumPy')
plt.xlabel("Matrix Number (1 = 3x3, ..., 8 = 384x384)")
plt.ylabel("Runtime (seconds)")
plt.title("Naive vs NumPy Matrix Multiplication Runtime")
plt.legend()
plt.grid()

plt.show()
```
