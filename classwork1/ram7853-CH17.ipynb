```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
k = 1.0
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0

plt.figure(figsize=(12, 6))

for h in h_values:
    #Euler coefficient matrix
    A = np.array([
        [1, h],
        [-h * k, 1]
    ])

    #Eigenvalues
    eigenvalues = np.linalg.eigvals(A)
    magnitudes = np.abs(eigenvalues)

    #Output stability info
    print(f"\nStep size h = {h}")
    print(f"Eigenvalues: {eigenvalues}")
    print(f"Magnitudes: {magnitudes}")

    if np.all(magnitudes < 1):
        print("Solution is Stable")
    else:
        print("Solution may not be Entirely Stable ;)")

    #Euler's Method
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(x_values)):
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

    #Plot solution
    plt.plot(x_values, y_values, label=f'h={h}')

#Final plot formatting
plt.xlabel('x')
plt.ylabel('y(x)')
plt.title(f'Euler Method Stability Analysis (k = {k})')
plt.legend()
plt.grid(True)

plt.show()
```
