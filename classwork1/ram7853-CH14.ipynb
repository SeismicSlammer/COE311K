```python
import numpy as np
import matplotlib.pyplot as plt

#Parameters
a = 0
b = 2 * np.pi
N = 50
x = np.linspace(a, b, N)
h = (b - a) / (N - 1)

def f(x):
    return np.sin(x)

#Forward Diff
df_forward = (f(x + h) - f(x)) / h

#Backward Diff
df_backward = (f(x) - f(x - h)) / h

#Central Diff
df_central = (f(x + h) - f(x - h)) / (2 * h)

#True derivative
df_true = np.cos(x)

#Plotting
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.plot(x, df_true, 'k-', label='Exact: $\cos(x)$', lw=2)
plt.plot(x, df_forward, 'r--', label='Forward Diff')
plt.plot(x, df_backward, 'b--', label='Backward Diff')
plt.plot(x, df_central, 'g:', label='Central Diff')
plt.title('Comparison of All Methods')
plt.legend()
plt.grid(True)

plt.subplot(2, 2, 2)
plt.plot(x, df_true, 'k-', label='Exact')
plt.plot(x, df_forward, 'r--', label='Forward')
plt.title('Forward Difference')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(x, df_true, 'k-', label='Exact')
plt.plot(x, df_backward, 'b--', label='Backward')
plt.title('Backward Difference')
plt.grid(True)
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x, df_true, 'k-', label='Exact')
plt.plot(x, df_central, 'g:', label='Central')
plt.title('Central Difference')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
