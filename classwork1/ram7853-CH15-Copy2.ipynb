```python
import numpy as np
import matplotlib.pyplot as plt

# Define the ODE
def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

# Analytical solution
def y_exact(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

# Step sizes
h_values = [0.1, 0.01, 0.001, 0.00001, 0.0000001]

t_start = 0
t_end = 30

plt.figure(figsize=(12, 6))

for h in h_values:
    print(f"\nRunning h = {h}")

    # WARNING: extremely large arrays for very small h
    t_values = np.arange(t_start, t_end + h, h)

    # Safety check (optional, prevents crashing)
    if len(t_values) > 5_000_000:
        print(f"Skipping h = {h} (too many points: {len(t_values)})")
        continue

    y_euler = np.zeros(len(t_values))
    y_euler[0] = 0

    # Euler Forward
    for i in range(1, len(t_values)):
        t_n = t_values[i - 1]
        y_n = y_euler[i - 1]

        y_euler[i] = y_n + h * f(t_n, y_n)

    # Plot Euler solution
    plt.plot(t_values, y_euler, label=f'Euler h={h}')

# Analytical solution (smooth reference)
t_fine = np.linspace(t_start, t_end, 10000)
plt.plot(t_fine, y_exact(t_fine), 'k', linewidth=2, label='Analytical')

# Plot formatting
plt.xlabel('t')
plt.ylabel('y(t)')
plt.title("Euler Forward vs Analytical (Stiff ODE)")
plt.legend()
plt.grid(True)

plt.show()
```
