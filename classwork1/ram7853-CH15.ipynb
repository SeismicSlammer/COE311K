```python
import numpy as np
import matplotlib.pyplot as plt

#Function
def f(x, y):
    return -2 * y

#Exact solution
def exact_sol(x):
    return np.exp(-2 * x)

x_start, x_end = 0, 2
step_sizes = np.arange(0.01, 1.05, 0.05)

plt.figure(figsize=(10, 6))

#Loop through each step size
for h in step_sizes:
    num_steps = int((x_end - x_start) / h)
    x_vals = np.arange(x_start, x_end + h, h)
    y_vals = np.zeros(len(x_vals))
    
    y_vals[0] = 1
    
    #Euler's Method
    for i in range(len(x_vals) - 1):
        y_vals[i+1] = y_vals[i] + h * f(x_vals[i], y_vals[i])
    
    #Error
    y_exact = exact_sol(x_vals)
    pointwise_error = np.abs(y_vals - y_exact)
    total_error = np.sum(pointwise_error)
    
    print(f"Step Size h={h:.2f} | Total Accumulated Error: {total_error:.4f}")
    
    #Plot current run
    plt.plot(x_vals, y_vals, label=f'h={h:.2f}')

#Plot
x_fine = np.linspace(x_start, x_end, 100)
plt.plot(x_fine, exact_sol(x_fine), 'k--', lw=2, label='Exact (e^-2x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method: Effect of Step Size on Error")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
